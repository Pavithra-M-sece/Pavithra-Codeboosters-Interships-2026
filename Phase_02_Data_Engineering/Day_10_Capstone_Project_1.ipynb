````python
# Function to analyze dataset using Groq-generated Pandas code

def analyze_data_with_groq(question, dataframe):

    """
    Purpose:
    --------
    Accepts a natural language question,
    generates Pandas code using Groq,
    executes the code,
    and returns JSON output.
    """

    print("Starting AI data analysis.")

    # Sample dataset records sent to Groq
    df_sample = dataframe.head().to_string()

    # Prompt sent to Groq
    prompt = f"""
You are an expert Python Data Analyst.

DataFrame name: df

Columns:
{dataframe.columns.tolist()}

Sample Data:
{df_sample}

Rules:
1. Generate only executable Python code.
2. Store final answer in variable named result.
3. Output must be JSON compatible.
4. No markdown.
5. No explanation.

Question:
{question}
"""

    try:

        print("Sending request to Groq.")

        response = client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=0
        )

        print("Response received from Groq.")

        generated_code = response.choices[0].message.content

        print("\nGenerated Code:")
        print("--------------------------------")
        print(generated_code)
        print("--------------------------------")

        # Remove markdown formatting if generated
        generated_code = generated_code.replace(
            "```python",
            ""
        )

        generated_code = generated_code.replace(
            "```",
            ""
        )

        print("Executing generated code.")

        # Local scope for execution
        local_scope = {
            "df": dataframe
        }

        # Execute code
        exec(
            generated_code,
            {},
            local_scope
        )

        print("Code executed successfully.")

        # Get result variable
        result = local_scope.get(
            "result",
            "No result generated."
        )

        # JSON output
        json_output = {
            "status": "success",
            "question": question,
            "result": result
        }

        print("\nJSON Output:")

        print(
            json.dumps(
                json_output,
                indent=4,
                default=str
            )
        )

        print("\nAnalysis completed successfully.")

        return json_output

    except Exception as e:

        error_output = {
            "status": "error",
            "question": question,
            "error_message": str(e)
        }

        print(
            json.dumps(
                error_output,
                indent=4
            )
        )

        return error_output
````

````python
# Calculate average math score

result = analyze_data_with_groq(
    "What is the average math score?",
    df
)

print("Query executed successfully.")
````
